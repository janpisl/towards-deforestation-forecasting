```python
import geopandas as gpd
import geopandas as gpd
import numpy as np
from shapely.geometry import box
import torch
```

```python

df = pd.read_csv("../data/annotation.csv")
df['geometry'] = gpd.GeoSeries.from_wkt(df['geometry'])
gdf = gpd.GeoDataFrame(df)
gdf.crs = 4326
```

```python
gdf['sampleid'] = range(len(gdf))
```

```python
def create_grid(bounds, cell_size, crs):
    xmin, ymin, xmax, ymax = bounds
    x_coords = np.arange(xmin, xmax, cell_size)
    y_coords = np.arange(ymin, ymax, cell_size)
    
    cells = [box(x, y, x + cell_size, y + cell_size) for x in x_coords for y in y_coords]
    
    return gpd.GeoDataFrame(geometry=cells, crs=crs)

```

```python
amazon_border = gpd.read_file('../data/terrabrasilis/amazon_biome_border/amazon_biome_border.shp')
amazon_geom = amazon_border.geometry.simplify(0.01).item()
```

```python


#Generate a grid of cells
grid = create_grid(gdf.total_bounds, cell_size=0.2, crs=gdf.crs)
#Drop cells not intersecting with target area
grid = grid.loc[grid.intersects(amazon_geom)].reset_index(drop=True)
```

```python
grid.shape
```

```python
#Randomly split cells into train, validation and test subsets

n_cells = len(grid)
indices = np.arange(n_cells)
np.random.shuffle(indices)

split1, split2 = int(0.6 * n_cells), int(0.8 * n_cells)

train = indices[:split1]
val = indices[split1:split2]
test = indices[split2:]


#grid.to_file('../data/tmp/grid.gpkg')
```

```python
#For validation and test sets, apply negative buffer to create distance between
#training and val/test data samples
#This is done to allow using a larger context for predicting each sample without causing data leakage
#The buffer size should be: patch_size + 2*context size; because up to 1/2 of a patch can be 
# crossing the cell border from each side; and the context size is added to patches from each side
# Here 0.05 degree is approx. 5km; patch size is 50px = 1.5km; 
#so context size can be approx. 1.75km on each side of the patch; 
#so the max patch size should be approx 5x5km; 167

buffer_size = -0.05

grid.loc[val, "geometry"] = grid.loc[val, "geometry"].buffer(buffer_size, cap_style=2)
grid.loc[test, "geometry"] = grid.loc[test, "geometry"].buffer(buffer_size, cap_style=2)
```

```python
import tqdm
```

```python
# Assign a cell id to samples

from shapely.strtree import STRtree
import numpy as np
from tqdm import tqdm

# Ensure geometries are proper Shapely objects
gdf_geoms = list(gdf.geometry)
tree = STRtree(gdf_geoms)

# Initialize the result array
cell_column = np.full(len(gdf), np.nan)

# Iterate over grid geometries
for idx, grid_geom in enumerate(tqdm(list(grid.geometry))):
    candidate_indices = tree.query(grid_geom)  # Returns an array of indices
    for candidate_idx in candidate_indices:
        candidate_geom = gdf_geoms[candidate_idx]
        if grid_geom.intersects(candidate_geom):
            cell_column[candidate_idx] = idx

gdf["cell"] = cell_column
```

```python
# Based on cell ids and the division of cells into train/val/test sets, 
# assign samples to the corresponding dataset split


gdf.loc[gdf.cell.isin(train), "split"] = 'train'
gdf.loc[gdf.cell.isin(val), "split"] = 'val'
gdf.loc[gdf.cell.isin(test), "split"] = 'test'


print(f"Train samples: {len(gdf.loc[gdf.split == 'train'])}")
print(f"Val samples: {len(gdf.loc[gdf.split == 'val'])}")
print(f"Test samples: {len(gdf.loc[gdf.split == 'test'])}")
print(f"Unassigned samples: {len(gdf.loc[~gdf.cell.notna()])}")
```

```python
gdf.cell
```

```python
# Select 1/5 of the full validation set to be the 'operational' validation set
gdf.loc[gdf.split == 'val', 'split']  = 'val_full' 
sample_val_ids = gdf.loc[gdf.split == 'val_full'].sample((len(gdf.loc[gdf.split == 'val_full'])//5)).sampleid
gdf.loc[gdf.sampleid.isin(sample_val_ids) , 'split'] = 'val'
```

```python
# Identify samples that only contain 0s in 2008-2018, and move them from the 'operational'
# validation set to the full one - because they are not very informative
val = gdf.loc[gdf.split == 'val']
empty_sequences_val_sampleid = val.loc[(val[['n_pixels_2008',
       'n_pixels_2009', 'n_pixels_2010', 'n_pixels_2011', 'n_pixels_2012',
       'n_pixels_2013', 'n_pixels_2014', 'n_pixels_2015', 'n_pixels_2016',
       'n_pixels_2017', 'n_pixels_2018']].sum(axis=1) == 0 )].sampleid

gdf.loc[gdf.sampleid.isin(empty_sequences_val_sampleid) , 'split'] = 'val_full'
```

```python
# Drop duplicate geometries
duplicates = gdf.duplicated(subset='geometry', keep=False)
gdf_drop_duplicates = gdf[~duplicates]
```

```python
# Drop samples that haven't been assigned to train/val/test; 
# This happens because of the gap between train, val, test splits
gdf_drop_unassigned = gdf_drop_duplicates[~gdf_drop_duplicates.split.isna()].reset_index(drop=True)
```

```python
import geopandas as gpd
import numpy as np

# Assume gdf is your GeoDataFrame of uniform-size square polygons
_gdf = _gdf.reset_index(drop=True)
square_area = _gdf.geometry.iloc[0].area
threshold = 0.9 * square_area

# Perform spatial join: each polygon with all others whose bounding boxes intersect
joined = gpd.sjoin(_gdf, _gdf, predicate="intersects", how="inner", lsuffix="1", rsuffix="2")

# Remove self-joins
joined = joined[joined.index1 != joined.index2]

# Compute intersection areas vectorized
intersections = _gdf.geometry.take(joined.index1.values).intersection(_gdf.geometry.take(joined.index2.values))
overlap_mask = intersections.area.values >= threshold

# Filter for high-overlap pairs
result = joined[overlap_mask][["index1", "index2"]].drop_duplicates()
```
